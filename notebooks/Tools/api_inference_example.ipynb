# SimpleDet API Example: Inference-oriented Workflow

This notebook shows how to build a detector and inspect the resulting pipeline
without requiring a full training loop.


```python
from simpledet.suite import build_detector, compile_detector_spec
from simpledet import create_project_inference_pipeline
```

```python
spec = build_detector(
    "deformable_detr",
    encoder="convnext_tiny.in12k_ft_in1k",
    decoder={"type": "DeformableDETRTransformerDecoder", "num_layers": 6},
    num_classes=1,
    in_channels=3,
)
model_cfg = compile_detector_spec(spec)
model_cfg["type"]
```

```python
pipeline = create_project_inference_pipeline(
    dataset_root="/path/to/data",
    result_folder="/tmp/simpledet-inference-example",
    categories=("wake",),
    in_channels=3,
    tif_channels_to_load=[1, 2, 3],
    model_cfg=model_cfg,
    max_epochs=1,
    batch_size=1,
)
pipeline.build()
```
