# Suite 00: Backbone Catalog And Auto-Adaptation

This notebook is the entry point to the new `simpledet.suite` API. It shows how to discover a backbone, build a detector spec, and compile it into an MMDetection-style model config without manually editing neck or head channels.


```python
from pprint import pprint

from simpledet._model_resolution import list_available_encoders
from simpledet.suite import build_detector, compile_detector_spec
```

```python
list_available_encoders(pattern="resnet*in1k")[:10]
```

```python
spec = build_detector(
    "retinanet",
    encoder="resnet18.a1_in1k",
    num_classes=2,
    in_channels=3,
)
model_cfg = compile_detector_spec(spec)
pprint(model_cfg["backbone"])
pprint(model_cfg["neck"])
pprint(model_cfg["bbox_head"])
```
