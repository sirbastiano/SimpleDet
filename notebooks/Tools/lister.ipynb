```python
import rasterio as rio 


def read_tif(path):
    with rio.open(path) as src:
        return src.read(10), src.profile
    
path = '/Data_large/marine/PythonProjects/MMDET/tgt_imgs/ASH_L0_09936_20190618_MultiLayer_mask_OK.tif'

img, profile = read_tif(path)

# write fuction to save_tif inot a tif. file new:
def save_tif(img, path, profile):
    with rio.open(path, 'w', **profile) as dst:
        dst.write(img, 10)
        
save_tif(img, '/Data_large/marine/PythonProjects/MMDET/tgt_imgs/SAV_ASH_L0_09936_20190618_MultiLayer_mask_OK.tif', profile)
```
