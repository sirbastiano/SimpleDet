# Suite 03: Transformer Detector Assembly

Build a transformer detector and inspect how the suite adapts ChannelMapper inputs, transformer embed dimensions, positional encodings, and class count.


```python
from pprint import pprint

from simpledet.suite import build_decoder, build_detector, build_neck, compile_detector_spec
```

```python
spec = build_detector(
    "deformable_detr",
    encoder="convnext_tiny.in12k_ft_in1k",
    neck=build_neck(name="ChannelMapper", out_channels=192, num_outs=4),
    decoder=build_decoder(embed_dims=192, num_queries=300),
    num_classes=5,
    in_channels=3,
)
model_cfg = compile_detector_spec(spec)
pprint(model_cfg["neck"])
pprint(model_cfg["encoder"])
pprint(model_cfg["decoder"])
pprint(model_cfg["bbox_head"])
pprint(model_cfg["positional_encoding"])
```
