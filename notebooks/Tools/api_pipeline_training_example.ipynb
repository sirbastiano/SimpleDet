# SimpleDet API Example: Pipeline Construction

This notebook demonstrates the higher-level public API built around `create_project_pipeline`
and `ObjectDetectionPipeline`.


```python
from pathlib import Path

from simpledet import create_project_pipeline, ObjectDetectionPipeline
from simpledet.suite import build_detector
```

```python
# 1) Build a detector spec with the suite API
detector_spec = build_detector(
    "retinanet",
    encoder="resnet50",
    num_classes=1,
    in_channels=3,
)
```

```python
# 2) Create a pipeline from a conventional dataset root
pipeline = create_project_pipeline(
    dataset_root="/path/to/data",
    categories=("wake",),
    in_channels=3,
    detector_spec=detector_spec,
    tif_channels_to_load=[1, 2, 3],
    resize=768,
    batch_size=4,
    max_epochs=12,
    learning_rate=1e-3,
    result_folder="/tmp/simpledet-example",
)
```

```python
# 3) Inspect and validate the operational setup
pipeline.summary()
pipeline.validate_paths(strict=False)
```
