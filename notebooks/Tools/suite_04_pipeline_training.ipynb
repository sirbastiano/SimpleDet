# Suite 04: Training Through `ObjectDetectionPipeline`

```python
from simpledet.api import ObjectDetectionPipeline
from simpledet.suite import build_detector
```

```python
detector_spec = build_detector(
    "retinanet",
    encoder="resnet18.a1_in1k",
    num_classes=2,
    in_channels=3,
)

pipeline = ObjectDetectionPipeline(
    detector_spec=detector_spec,
    data_folder="/path/to/dataset_root",
    result_folder="/path/to/results",
    config_folder="simpledet/simpledet/src",
    data_prefix="imgs/",
    annot_file_train="/path/to/annotations/train.json",
    annot_file_val="/path/to/annotations/val.json",
    annot_file_test="/path/to/annotations/test.json",
    tif_channels_to_load=[1, 1, 1],
    in_channels=3,
    categories=("object", "background_like"),
    resize=768,
    batch_size=2,
    learning_rate=0.001,
    max_epochs=10,
)

# pipeline.build()
# pipeline.train()
```
