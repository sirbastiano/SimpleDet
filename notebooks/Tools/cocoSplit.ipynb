```python
import json
from copy import deepcopy

def get_coco(annotation_file):
    """
    Load a COCO dataset from a JSON annotation file.

    Parameters:
    annotation_file (str): Path to the COCO annotation file.

    Returns:
    dict: Parsed COCO dataset as a dictionary.
    """
    with open(annotation_file, 'r') as f:
        data = json.load(f)
    return data

def filter_annotations_by_filename(coco_data, filenames):
    """
    Filter annotations in a COCO dataset to include only those matching the given image filenames.

    Parameters:
    coco_data (dict): The COCO dataset as a dictionary.
    filenames (set): A set of image filenames to filter annotations for.

    Returns:
    dict: The filtered COCO dataset containing only the specified filenames and corresponding annotations.
    """
    # Filter images to only include those with filenames in the set
    filtered_images = [img for img in coco_data['images'] if img['file_name'] in filenames]
    
    # Get the image IDs corresponding to the filtered images
    filtered_image_ids = {img['id'] for img in filtered_images}
    
    # Filter annotations to include only those associated with the filtered image IDs
    filtered_annotations = [ann for ann in coco_data['annotations'] if ann['image_id'] in filtered_image_ids]
    
    # Create a new COCO dataset dictionary with the filtered data
    filtered_coco = deepcopy(coco_data)
    filtered_coco['images'] = filtered_images
    filtered_coco['annotations'] = filtered_annotations

    return filtered_coco

# Load the datasets
train_coco = get_coco('/Data_large/marine/Datasets/V2RAW_SRC/annotations/train.json')
val_coco = get_coco('/Data_large/marine/Datasets/V2RAW_SRC/annotations/val.json')
test_coco = get_coco('/Data_large/marine/Datasets/V2RAW_SRC/annotations/test.json')

combined_coco = get_coco('/Data_large/marine/Datasets/VDS2Raw/annotations/combined_coco.json')

# Get the filenames for each split
train_filenames = {img['file_name'] for img in train_coco['images']}
val_filenames = {img['file_name'] for img in val_coco['images']}
test_filenames = {img['file_name'] for img in test_coco['images']}

# Filter the combined COCO dataset annotations based on these filenames
train_filtered_coco = filter_annotations_by_filename(combined_coco, train_filenames)
val_filtered_coco = filter_annotations_by_filename(combined_coco, val_filenames)
test_filtered_coco = filter_annotations_by_filename(combined_coco, test_filenames)

# Save the filtered datasets to new files
with open('/Data_large/marine/Datasets/VDS2Raw/annotations/filtered_train.json', 'w') as f:
    json.dump(train_filtered_coco, f, indent=4)

with open('/Data_large/marine/Datasets/VDS2Raw/annotations/filtered_val.json', 'w') as f:
    json.dump(val_filtered_coco, f, indent=4)

with open('/Data_large/marine/Datasets/VDS2Raw/annotations/filtered_test.json', 'w') as f:
    json.dump(test_filtered_coco, f, indent=4)
```
