```python
import os
import json 
import pandas as pd
from pathlib import Path


def read_json2(json_path):    
    with open(json_path, 'r') as file:
        # Read the content of the file
        content = file.read()
        # Replace single quotes with double quotes
        content = content.replace("'", '"')
        # Load the JSON data
        data = json.loads(content)
        precision = data.get('precision')
        recall = data.get('recall')
        f1_score = data.get('f1_score')
        
    seed, LR, BS = json_path.parent.name.split('_')[0], json_path.parent.name.split('_')[4], json_path.parent.name.split('_')[2]   
    
    return {'seed': seed, 'LR': LR, 'BS': BS, 'precision': precision, 'recall': recall, 'f1_score': f1_score}

# subfolders = [x for x in Path('/Data_large/marine/PythonProjects/MMDET/checkpoints/Sentinel/Special/BS_4/LR_0.0005/IMG_2048/').iterdir() if x.is_dir()]
# for fold in subfolders:
#     results = [x for x in fold.glob("**/*/*SIoU*.json")]
#     print(len(results))
#     appended = []
#     for item in results:
#         x = read_json2(item)
#         appended.append(x)

#     print(fold)
#     print(pd.DataFrame(appended).mean())
#     print(pd.DataFrame(appended).std())
#     print('\n')

```

```python
folder = '/Data_large/marine/PythonProjects/MMDET/checkpoints/VENuS/Single'

subfolders = [x for x in Path(folder).iterdir() if x.is_dir()]


results = {i:[] for i in range(1,13,1)}

for idx, sub in enumerate(subfolders):
    print(sub)
    i = int(sub.name.split('b')[-1])

    file_jsons = [x for x in sub.glob("**/*/*SIoU_40*.json")]

    results[i] = pd.DataFrame([read_json2(item) for item in file_jsons])
    
```
